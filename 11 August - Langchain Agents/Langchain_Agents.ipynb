```python
df = pd.read_csv(r'E:\GenAI\11 August\titanic_ dataset_final.csv')
print(df.head())
print(df.shape)
```

# Step 5: Single DataFrame Example

# Agent can interact with Single DataFrame

```python
llm  = ChatOpenAI()
```

```python
llm = ChatOpenAI(temperature=1,model="gpt-5.4-mini")
```

```python
agent = create_pandas_dataframe_agent(
    llm,
    df,
    verbose=True,
    agent_type="openai-tools",
    allow_dangerous_code=True
)
agent.run("How many passengers were there in the dataset?")
```

# Step 6 : Multi DataFrame Example

# Agent can also interact with Multiple DataFrames passed in a list

```python
df.info()
```

```python
df1 = df.copy()
```

```python
df1["Age"] = df1["Age"].fillna(df1["Age"].mean())
```

```python
agent = create_pandas_dataframe_agent(
    llm,
    [df, df1],
    verbose=True,
    allow_dangerous_code=True
)
```
