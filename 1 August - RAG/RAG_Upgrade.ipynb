```python
import os
from langchain_community.document_loaders import UnstructuredPDFLoader
from langchain_text_splitters import CharacterTextSplitter
from langchain_community.embeddings import HuggingFaceBgeEmbeddings
from langchain_chroma import Chroma
from langchain_groq import ChatGroq
from langchain_classic.chains import RetrievalQA
from langchain_community.document_loaders import UnstructuredFileLoader
```

```python
#Fetch the PDF From the URL
import requests
url= "https://www.mathcentre.ac.uk/resources/uploaded/mc-ty-introfns-2009-1.pdf"
response =requests.get(url)
```

```python
#Save the PDF to local file
with open ("python_inbuildfunction.pdf","wb") as f:
  f.write(response.content)
```

```python
#Loading the document
loader =  UnstructuredFileLoader("python_inbuildfunction.pdf")
```

```python
!pip install unstructured
```

```python
# !pip install "unstructured[pdf]"
```

```python
documents = loader.load()
documents
```

```python
text_splitter = CharacterTextSplitter(chunk_size=2000, chunk_overlap=400)
texts = text_splitter.split_documents(documents)
```

```python
type(texts)
```

```python
texts[4]
```

```python
embeddings = HuggingFaceBgeEmbeddings()
```

```python
persist_directory ="vector_db"
```

```python
vectordb = Chroma.from_documents(
    documents = texts,
    embedding = embeddings,
    persist_directory = persist_directory
)
```

```python
#retriever
retriver = vectordb.as_retriever()
```

```python
#llm from groq
llm = ChatGroq(
    model="llama-3.3-70b-versatile",
    temperature =0

)
```

```python
!pip uninstall -y langchain langchain-core langchain-community langchain-text-splitters langchain-classic
```

```python
qa_chain = RetrivalQA.from_chain_type(
    llm=llm,
    chain_type="stuff",
    retriever=retriver,
    return_source_documents=True
)
```

```python
#invoke the qa chain and get a response for user query
query = "what are the funnction from this pdf"
response = qa_chain.invoke({"query":query})
```

```python
print(response)
```

```python
print(response["result"])
```

```python
print(response["source_documents"][0].metadata["source"])
```
